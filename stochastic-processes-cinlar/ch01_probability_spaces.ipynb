## Event operations

Given an event $A \subseteq \Omega$:

### Complement

$$
A^c = \{\omega \in \Omega : \omega \notin A\}.
$$

This is the event that $A$ does not occur.

### Union

$$
A \cup B = \{\omega \in \Omega : \omega \in A \text{ or } \omega \in B\}.
$$

This is the event that at least one of $A$ or $B$ occurs.

### Intersection

$$
A \cap B = \{\omega \in \Omega : \omega \in A \text{ and } \omega \in B\}.
$$

This is the event that both occur.

### De Morgan's laws

$$
(A \cup B)^c = A^c \cap B^c,
\qquad
(A \cap B)^c = A^c \cup B^c.
$$

Mental model: complements flip “or” into “and,” and “and” into “or.”


```python
Omega = set(range(10))
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
Ac = Omega - A
Bc = Omega - B
print("A ∪ B =", sorted(A | B))
print("A ∩ B =", sorted(A & B))
print("(A ∪ B)^c =", sorted(Omega - (A | B)))
print("A^c ∩ B^c =", sorted(Ac & Bc))
print("(A ∩ B)^c =", sorted(Omega - (A & B)))
print("A^c ∪ B^c =", sorted(Ac | Bc))
```

## Disjoint events

Two events $A$ and $B$ are **disjoint** if they cannot both occur:

$$
A \cap B = \varnothing.
$$

A family $A_1,A_2,\ldots$ is disjoint if every pair is disjoint.

Union and intersection extend to sequences:

$$
\bigcup_{i=1}^{\infty} A_i
$$

is the event that at least one $A_i$ occurs, and

$$
\bigcap_{i=1}^{\infty} A_i
$$

is the event that all $A_i$ occur.


## Definition 1.6: probability measure

Let $\Omega$ be a sample space. A function $P$ assigning a number $P(A)$ to each event $A$ is a **probability measure** when:

1. For every event $A$,

$$
0 \le P(A) \le 1.
$$

2. The whole sample space has probability one:

$$
P(\Omega)=1.
$$

3. For any disjoint sequence $A_1,A_2,\ldots$,

$$
P\left(\bigcup_{i=1}^{\infty} A_i\right)
= \sum_{i=1}^{\infty} P(A_i).
$$

The triple $(\Omega, \mathcal F, P)$ is a probability space. In this chapter, events are treated as the subsets of $\Omega$ under discussion.

Mental model: probability is an area/size function on events. The key rule is that non-overlapping pieces add.


## Proposition 1.7: finite additivity for disjoint events

If $A_1,\ldots,A_n$ are disjoint, then

$$
P(A_1 \cup \cdots \cup A_n)
= P(A_1)+\cdots+P(A_n).
$$

### Proof

First prove that $P(\varnothing)=0$.

Since $\Omega \cup \varnothing \cup \varnothing \cup \cdots = \Omega$ is a disjoint union, countable additivity gives

$$
P(\Omega)=P(\Omega)+P(\varnothing)+P(\varnothing)+\cdots.
$$

Because $P(\Omega)=1$, the only possibility is

$$
P(\varnothing)=0.
$$

Now define

$$
A_{n+1}=A_{n+2}=\cdots=\varnothing.
$$

The infinite sequence $A_1,A_2,\ldots$ is disjoint. Therefore,

$$
P\left(\bigcup_{i=1}^{n}A_i\right)
= P\left(\bigcup_{i=1}^{\infty}A_i\right)
= \sum_{i=1}^{\infty}P(A_i)
= \sum_{i=1}^{n}P(A_i).
$$


## Proposition 1.8: monotonicity

If $A \subseteq B$, then

$$
P(A) \le P(B).
$$

### Proof

If $A \subseteq B$, then $B$ can be split into two disjoint pieces:

$$
B = A \cup (A^c \cap B).
$$

By finite additivity,

$$
P(B)=P(A)+P(A^c \cap B).
$$

Since probabilities are nonnegative,

$$
P(A^c \cap B) \ge 0,
$$

so

$$
P(B) \ge P(A).
$$


## Proposition 1.9: complement rule

For any event $A$,

$$
P(A^c)=1-P(A).
$$

### Proof

The events $A$ and $A^c$ are disjoint, and

$$
A \cup A^c = \Omega.
$$

By finite additivity,

$$
P(A)+P(A^c)=P(\Omega)=1.
$$

Therefore,

$$
P(A^c)=1-P(A).
$$

Mental model: if an event is hard to compute directly, compute the probability that it fails and subtract from $1$.


## Theorem 1.10: partition formula / total probability over events

If $B_1,B_2,\ldots$ are disjoint events with

$$
\bigcup_{i=1}^{\infty}B_i = \Omega,
$$

then for every event $A$,

$$
P(A)=\sum_{i=1}^{\infty}P(A \cap B_i).
$$

### Proof

Because the $B_i$ cover $\Omega$,

$$
A = A \cap \Omega
= A \cap \left(\bigcup_{i=1}^{\infty} B_i\right)
= \bigcup_{i=1}^{\infty}(A \cap B_i).
$$

The events $A \cap B_i$ are disjoint because the $B_i$ are disjoint. By countable additivity,

$$
P(A)=\sum_{i=1}^{\infty}P(A \cap B_i).
$$

Mental model: split the universe into cases $B_i$, compute $A$ inside each case, then add.


```python
# A finite partition illustration.
Omega = np.arange(1, 13)       # outcomes 1..12, equally likely
A = set(x for x in Omega if x % 2 == 0)      # even numbers
B1 = set(x for x in Omega if x <= 4)
B2 = set(x for x in Omega if 5 <= x <= 8)
B3 = set(x for x in Omega if x >= 9)
partition = [B1, B2, B3]

def P(event):
    return len(event) / len(Omega)

lhs = P(A)
rhs = sum(P(A & B) for B in partition)
lhs, rhs
```

## Proposition 1.11: continuity from below

Let $A_1,A_2,\ldots$ be an increasing sequence of events:

$$
A_1 \subseteq A_2 \subseteq A_3 \subseteq \cdots.
$$

Let

$$
A = \bigcup_{n=1}^{\infty}A_n.
$$

Then

$$
P(A)=\lim_{n\to\infty}P(A_n).
$$

### Proof

Define disjoint increments:

$$
B_1=A_1,
\qquad
B_n = A_n \cap A_{n-1}^c \quad (n\ge2).
$$

Then the $B_n$ are disjoint and

$$
\bigcup_{i=1}^{n}B_i = A_n,
\qquad
\bigcup_{i=1}^{\infty}B_i = A.
$$

By finite additivity,

$$
P(A_n)=\sum_{i=1}^{n}P(B_i).
$$

Taking limits,

$$
\lim_{n\to\infty}P(A_n)
=\sum_{i=1}^{\infty}P(B_i)
=P(A).
$$


```python
# Increasing events: A_n = {1, ..., n} in a geometric distribution on positive integers.
p = 0.3
n = np.arange(1, 21)
P_An = 1 - (1-p)**n
pd.DataFrame({"n": n, "P(A_n)": P_An}).head(), P_An[-1]
```

## Corollary 1.12: continuity from above

Let $A_1,A_2,\ldots$ be a decreasing sequence of events:

$$
A_1 \supseteq A_2 \supseteq A_3 \supseteq \cdots.
$$

Let

$$
A = \bigcap_{n=1}^{\infty}A_n.
$$

Then

$$
P(A)=\lim_{n\to\infty}P(A_n).
$$

### Proof

The complements are increasing:

$$
A_1^c \subseteq A_2^c \subseteq A_3^c \subseteq \cdots.
$$

Also,

$$
\bigcup_{n=1}^{\infty}A_n^c
= \left(\bigcap_{n=1}^{\infty}A_n\right)^c
= A^c.
$$

By continuity from below,

$$
P(A^c)=\lim_{n\to\infty}P(A_n^c).
$$

Using complements,

$$
1-P(A)=\lim_{n\to\infty}(1-P(A_n)).
$$

Therefore,

$$
P(A)=\lim_{n\to\infty}P(A_n).
$$

## Example 2.4: vehicle acceleration, velocity, and position

Experiment: observe acceleration $\omega(t)$ of a vehicle during the first 60 seconds.

The sample space consists of continuous functions on $[0,60]$.

For each $t \in [0,60]$, define

$$
X_t(\omega)=\omega(t),
$$

$$
Y_t(\omega)=\int_0^t \omega(s)\,ds,
$$

$$
Z_t(\omega)=\int_0^t Y_s(\omega)\,ds
= \int_0^t\int_0^s \omega(u)\,du\,ds.
$$

Interpretation:

- $X_t$ is acceleration at time $t$.
- $Y_t$ is velocity at time $t$.
- $Z_t$ is position at time $t$.

The collection $\{X_t:0\le t\le 60\}$ is a **continuous-time stochastic process**.


```python
# Simulate one smooth acceleration path and integrate it numerically.
t = np.linspace(0, 60, 601)
acc = 0.15*np.sin(2*np.pi*t/20) + 0.03*rng.normal(size=len(t))
dt = t[1] - t[0]
vel = np.cumsum(acc) * dt
pos = np.cumsum(vel) * dt

plt.figure(figsize=(8, 4))
plt.plot(t, acc, label="acceleration X_t")
plt.plot(t, vel, label="velocity Y_t")
plt.plot(t, pos/60, label="position Z_t / 60")
plt.xlabel("time t")
plt.legend()
plt.title("One simulated path: acceleration, velocity, and position")
plt.show()
```

## Stochastic processes

A family of random variables

$$
X = \{X_t : t \in T\}
$$

is called a **stochastic process**.

- $T=\{0,1,2,\ldots\}$ gives a discrete-time process.
- $T=[0,\infty)$ gives a continuous-time process.

For fixed $\omega$, the function

$$
t \mapsto X_t(\omega)
$$

is called a **sample path**.

## Example 2.6: arrival counting process

An outcome is a nondecreasing sequence

## Events induced by random variables

For a real-valued random variable $X$, sets like

$$
\{X \le a\}
$$

mean

$$
\{\omega \in \Omega : X(\omega) \le a\}.
$$

If $a\le b$, then

$$
\{X\le a\}\subseteq\{X\le b\}.
$$

By monotonicity,

$$
P\{X\le a\}\le P\{X\le b\}.
$$


## Distribution function

The **distribution function** of a real random variable $X$ is

$$
\varphi(b)=P\{X\le b\},\qquad -\infty<b<\infty.
$$

It satisfies:

1. $\varphi$ is nondecreasing.
2. $\varphi$ is right-continuous.
3. $\lim_{b\to -\infty}\varphi(b)=0$.
4. $\lim_{b\to +\infty}\varphi(b)=1$.

### Why it is nondecreasing

If $a\le b$, then

$$
\{X\le a\}\subseteq\{X\le b\}.
$$

Thus

$$
P\{X\le a\}\le P\{X\le b\}.
$$

### Why it is right-continuous

Let $b_n \downarrow b$. Then

$$
\{X\le b_n\}\downarrow \{X\le b\}.
$$

By continuity from above,

$$
P\{X\le b_n\}\to P\{X\le b\}.
$$

### Why the limits are 0 and 1

As $b\to +\infty$, the events $\{X\le b\}$ increase to $\Omega$, so probabilities increase to $1$.

As $b\to -\infty$, the events $\{X\le b\}$ decrease to $\varnothing$, so probabilities decrease to $0$.

## Example 2.17: generating a random variable with a given distribution

Suppose a computer can generate a uniform random number

$$
\omega \in [0,1].
$$

Let $\Omega=[0,1]$, and let $P$ be ordinary length probability on $[0,1]$.

Suppose $\varphi$ is continuous and strictly increasing, with range from $0$ to $1$.

For each $\omega\in[0,1]$, define $X(\omega)=a$ where

$$
\varphi(a)=\omega.
$$

Equivalently,

$$
X(\omega)=\varphi^{-1}(\omega).
$$

Then

$$
P\{X\le t\}=\varphi(t).
$$

### Proof

Because $\varphi$ is increasing,

$$
X(\omega)\le t
\iff
\varphi^{-1}(\omega)\le t
\iff
\omega\le \varphi(t).
$$

Thus

$$
P\{X\le t\}=P\{\omega\le \varphi(t)\}=\varphi(t),
$$

because $\omega$ is uniform on $[0,1]$.

This is the **inverse transform method** for simulation.


```python
# Example: exponential distribution with rate lambda.
lam = 2.0
u = rng.uniform(size=10_000)
X = -np.log(1-u) / lam  # inverse CDF

x = np.linspace(0, 4, 200)
empirical_F = np.array([(X <= xi).mean() for xi in x])
theoretical_F = 1 - np.exp(-lam*x)

plt.figure(figsize=(7, 4))
plt.plot(x, empirical_F, label="empirical CDF")
plt.plot(x, theoretical_F, label="theoretical CDF")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.legend()
plt.title("Inverse transform sampling: exponential distribution")
plt.show()
```

## Example 2.18: thrust correction and a transformed random variable

A mid-course trajectory correction calls for a velocity increase of $135$ ft/sec.

The spacecraft engine provides constant acceleration of $15$ ft/sec$^2$.

Let $T$ be the actual firing time. The distribution function of $T$ is

$$
P\{T\le t\}
=
\begin{cases}
0, & t<9,\\
1-e^{-3(t-9)}, & t\ge 9.
\end{cases}
$$

The intended burn time is $9$ seconds, because

$$
15\cdot 9=135.
$$

Let $X$ be the actual velocity increase:

$$
X=15T.
$$

Then

$$
P\{X\le b\}=P\{15T\le b\}=P\{T\le b/15\}.
$$

Therefore,

$$
P\{X\le b\}
=
\begin{cases}
0, & b<135,\\
1-e^{-3(b/15-9)}, & b\ge 135.
\end{cases}
$$

Equivalently,

$$
P\{X\le b\}=1-e^{-(b-135)/5},\qquad b\ge 135.
$$


```python
b = np.linspace(100, 180, 300)
F_X = np.where(b < 135, 0, 1 - np.exp(-(b-135)/5))
plt.figure(figsize=(7, 4))
plt.plot(b, F_X)
plt.xlabel("velocity increase b")
plt.ylabel("P{X≤b}")
plt.title("Distribution of actual velocity increase X=15T")
plt.show()
```

## Discrete random variables and probability distributions

Let $X$ be a discrete random variable taking values in a countable set $E$.

For each $a\in E$, define

$$
\pi(a)=P\{X=a\}.
$$

Then

$$
\pi(a)\ge 0,
\qquad
\sum_{a\in E}\pi(a)=1.
$$

The collection $\{\pi(a):a\in E\}$ is the **probability distribution** of $X$.

For real-valued discrete $X$, $\pi$ is sometimes called the **probability mass function**.


## Joint distributions

For random variables $X_1,\ldots,X_n$, their joint distribution is specified by probabilities such as

$$
P\{X_1=a_1,\ldots,X_n=a_n\}
$$

or, for real-valued variables,

$$
P\{X_1\le b_1,\ldots,X_n\le b_n\}.
$$

The joint distribution is the full probabilistic description of the vector

$$
(X_1,\ldots,X_n).
$$


## Definition 2.19: independence of discrete random variables

Discrete random variables $X_1,\ldots,X_n$ are **independent** if for all values $a_1,\ldots,a_n$,

$$
P\{X_1=a_1,\ldots,X_n=a_n\}
= P\{X_1=a_1\}\cdots P\{X_n=a_n\}.
$$

For real random variables, independence means that for all thresholds $b_1,\ldots,b_n$,

$$
P\{X_1\le b_1,
\ldots,
X_n\le b_n\}
= P\{X_1\le b_1\}\cdots P\{X_n\le b_n\}.
$$

Mental model: knowing one variable gives no information about the others.


## Example 2.22: independent geometric-like random variables

Let $X$ and $Y$ be discrete random variables taking values in $\{1,2,3,\ldots\}$ with joint distribution

$$
P\{X=m,Y=n\}=0.64(0.2)^{m+n-2},
\qquad m,n=1,2,\ldots.
$$

Compute the marginal distribution of $X$:

$$
P\{X=m\}
=\sum_{j=1}^{\infty}P\{X=m,Y=j\}
=\sum_{j=1}^{\infty}0.64(0.2)^{m+j-2}.
$$

Factor out the terms not depending on $j$:

$$
P\{X=m\}
=0.64(0.2)^{m-1}\sum_{j=1}^{\infty}(0.2)^{j-1}.
$$

Since

$$
\sum_{j=1}^{\infty}(0.2)^{j-1}=\frac{1}{1-0.2}=1.25,
$$

we get

$$
P\{X=m\}=0.8(0.2)^{m-1}.
$$

Similarly,

$$
P\{Y=n\}=0.8(0.2)^{n-1}.
$$

Therefore,

$$
P\{X=m,Y=n\}
= P\{X=m\}P\{Y=n\}.
$$

Thus $X$ and $Y$ are independent.


```python
# Numerically check the first few probabilities.
def joint_222(m, n):
    return 0.64*(0.2)**(m+n-2)

def marg(m):
    return 0.8*(0.2)**(m-1)

rows=[]
for m in range(1,5):
    for n in range(1,5):
        rows.append({
            "m": m, "n": n,
            "joint": joint_222(m,n),
            "product_of_marginals": marg(m)*marg(n)
        })
pd.DataFrame(rows).head(8)
```

## Example 2.23: a joint distribution that is not independent

Let $X$ and $Y$ take values in $\{0,1,2,\ldots\}$ with joint distribution

$$
P\{X=m,Y=n\}=
\begin{cases}
\dfrac{e^{-7}4^m3^{n-m}}{m!(n-m)!}, & m=0,1,\ldots,n,\\
0, & \text{otherwise.}
\end{cases}
$$

For $m\in\mathbb N$, compute the marginal distribution of $X$:

$$
P\{X=m\}
=\sum_{j=m}^{\infty}
\frac{e^{-7}4^m3^{j-m}}{m!(j-m)!}.
$$

Let $k=j-m$:

$$
P\{X=m\}
=\frac{e^{-7}4^m}{m!}
\sum_{k=0}^{\infty}\frac{3^k}{k!}.
$$

Since

$$
\sum_{k=0}^{\infty}\frac{3^k}{k!}=e^3,
$$

we get

$$
P\{X=m\}=\frac{e^{-4}4^m}{m!}.
$$

Similarly,

$$
P\{Y=n\}
=\sum_{i=0}^{n}
\frac{e^{-7}4^i3^{n-i}}{i!(n-i)!}.
$$

Using the binomial theorem,

$$
\sum_{i=0}^{n}\frac{4^i3^{n-i}}{i!(n-i)!}
=\frac{1}{n!}\sum_{i=0}^{n}\binom{n}{i}4^i3^{n-i}
=\frac{7^n}{n!}.
$$

Therefore,

$$
P\{Y=n\}=\frac{e^{-7}7^n}{n!}.
$$

But

$$
P\{X=m,Y=n\}\ne P\{X=m\}P\{Y=n\}
$$

in general, so $X$ and $Y$ are not independent.

Interpretation: $X$ is like part of $Y$; the condition $m\le n$ already creates dependence.


```python
def joint_223(m, n):
    if 0 <= m <= n:
        return exp(-7) * 4**m * 3**(n-m) / (factorial(m)*factorial(n-m))
    return 0.0

def px_223(m):
    return exp(-4)*4**m/factorial(m)

def py_223(n):
    return exp(-7)*7**n/factorial(n)

m, n = 2, 5
joint_223(m,n), px_223(m)*py_223(n)
```

## Example 2.25: $X$ and $Y-X$ are independent in Example 2.23

Using the same joint distribution as Example 2.23, define

$$
Z=Y-X.
$$

For $m,k\in\mathbb N$,

$$
P\{X=m,Y-X=k\}=P\{X=m,Y=m+k\}.
$$

Using the joint distribution,

$$
P\{X=m,Y-X=k\}
=\frac{e^{-7}4^m3^k}{m!k!}.
$$

From Example 2.23,

$$
P\{X=m\}=\frac{e^{-4}4^m}{m!}.
$$

Also,

$$
P\{Y-X=k\}
=\sum_{i=0}^{\infty}P\{X=i,Y-X=k\}
=\sum_{i=0}^{\infty}\frac{e^{-7}4^i3^k}{i!k!}.
$$

Therefore,

$$
P\{Y-X=k\}
=\frac{e^{-7}3^k}{k!}\sum_{i=0}^{\infty}\frac{4^i}{i!}
=\frac{e^{-7}3^k}{k!}e^4
=\frac{e^{-3}3^k}{k!}.
$$

So

$$
P\{X=m,Y-X=k\}
=\left(\frac{e^{-4}4^m}{m!}\right)
\left(\frac{e^{-3}3^k}{k!}\right)
= P\{X=m\}P\{Y-X=k\}.
$$

Thus $X$ and $Y-X$ are independent.

Mental model: the joint law describes $Y$ as a sum of two independent Poisson parts: $X$ and $Y-X$.


# 3. Conditional probability

Let $A$ and $B$ be two events with $P(B)>0$.

The **conditional probability** of $A$ given $B$ is

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}.
$$

This defines a new probability measure on the same event space, with $B$ treated as the new universe.

Mental model: conditioning means restrict attention to worlds where $B$ happened, then renormalize probabilities so $B$ has total mass one.


## Properties of conditional probability

For fixed $B$ with $P(B)>0$:

1. Nonnegativity:

$$
0\le P(A\mid B)\le 1.
$$

2. Normalization:

$$
P(\Omega\mid B)=1.
$$

3. Countable additivity for disjoint $A_1,A_2,\ldots$:

$$
P\left(\bigcup_i A_i \mid B\right)
=\sum_i P(A_i\mid B).
$$

### Proof sketch

Each property follows from the definition

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}
$$

and the corresponding property of $P$.


## Theorem 3.3: total probability formula

If $B_1,B_2,\ldots$ are disjoint events with

$$
\bigcup_{i=1}^{\infty}B_i=\Omega,
$$

then for any event $A$,

$$
P(A)=\sum_{i=1}^{\infty}P(A\mid B_i)P(B_i).
$$

### Proof

By the partition formula,

$$
P(A)=\sum_{i=1}^{\infty}P(A\cap B_i).
$$

By the definition of conditional probability,

$$
P(A\cap B_i)=P(A\mid B_i)P(B_i).
$$

Substitute this into the sum:

$$
P(A)=\sum_{i=1}^{\infty}P(A\mid B_i)P(B_i).
$$


## Corollary 3.4: Bayes' formula

If $B_1,B_2,\ldots$ are disjoint events with union $\Omega$, then for any event $A$ with $P(A)>0$,

$$
P(B_j\mid A)
=
\frac{P(A\mid B_j)P(B_j)}{
\sum_i P(A\mid B_i)P(B_i)}.
$$

### Proof

Start with conditional probability:

$$
P(B_j\mid A)=\frac{P(A\cap B_j)}{P(A)}.
$$

Rewrite the numerator:

$$
P(A\cap B_j)=P(A\mid B_j)P(B_j).
$$

Rewrite the denominator by total probability:

$$
P(A)=\sum_i P(A\mid B_i)P(B_i).
$$

Substitute both into the first equation.

Mental model: Bayes' formula updates the probability of a cause/hypothesis $B_j$ after observing evidence $A$.


```python
# Simple Bayes illustration: disease testing.
prevalence = 0.01
sensitivity = 0.95         # P(+ | disease)
false_positive = 0.05      # P(+ | no disease)

p_positive = sensitivity*prevalence + false_positive*(1-prevalence)
posterior = sensitivity*prevalence / p_positive
posterior
```

## Example 3.5: first and second heads in coin tossing

Toss a coin repeatedly. Let

- $X$ be the trial number on which the first head occurs.
- $Y$ be the trial number on which the second head occurs.

Let

$$
q=1-p,
$$

where $p$ is the probability of heads.

For $m=1,2,\ldots$ and $n=2,3,\ldots$,

$$
P\{X=m,Y=n\}
=
\begin{cases}
q^{m-1}p\,q^{n-m-1}p = p^2q^{n-2}, & m<n,\\
0, & \text{otherwise.}
\end{cases}
$$

The marginal distribution of $X$ is

$$
P\{X=m\}=\sum_{j=m+1}^{\infty}P\{X=m,Y=j\}
=\sum_{j=m+1}^{\infty}p^2q^{j-2}.
$$

Using the geometric series,

$$
P\{X=m\}=pq^{m-1}.
$$

The marginal distribution of $Y$ is

$$
P\{Y=n\}=\sum_{i=1}^{n-1}P\{X=i,Y=n\}
=\sum_{i=1}^{n-1}p^2q^{n-2}
=(n-1)p^2q^{n-2}.
$$

### Conditional distribution of $X$ given $Y=n$

For $m<n$,

$$
P\{X=m\mid Y=n\}
=\frac{P\{X=m,Y=n\}}{P\{Y=n\}}
=\frac{p^2q^{n-2}}{(n-1)p^2q^{n-2}}
=\frac{1}{n-1}.
$$

Given that the second head occurred at trial $n$, the first head is equally likely to have occurred at any of trials $1,2,\ldots,n-1$.

### Conditional distribution of $Y$ given $X=m$

For $n=m+1,m+2,\ldots$,

$$
P\{Y=n\mid X=m\}
=\frac{P\{X=m,Y=n\}}{P\{X=m\}}
=\frac{p^2q^{n-2}}{pq^{m-1}}
=pq^{n-m-1}.
$$

Given the first head occurred at trial $m$, the waiting time to the second head is again geometric.

### Distribution of the gap $Y-X$

For $k=1,2,\ldots$,

$$
P\{Y-X=k\mid X=m\}
=P\{Y=m+k\mid X=m\}
=pq^{k-1}.
$$

This does not depend on $m$. Therefore, $Y-X$ is independent of $X$, and $Y-X$ has the same distribution as $X$.


```python
p = 0.4
q = 1-p
n = 8
conditional = {m: 1/(n-1) for m in range(1, n)}
pd.Series(conditional, name=f"P(X=m | Y={n})")
```

## Example 3.8: memoryless telephone conversation time

Suppose the length $X$ of a telephone conversation has distribution function

$$
P\{X\le t\}=1-e^{-0.03t},\qquad t\ge 0.
$$

A conversation has already gone on for $30$ minutes. What is the probability that it continues for at least another $20$ minutes?

We want

$$
P\{X>50\mid X>30\}.
$$

Using conditional probability,

$$
P\{X>50\mid X>30\}
=\frac{P\{X>50, X>30\}}{P\{X>30\}}.
$$

Since $\{X>50\}\subseteq\{X>30\}$,

$$
P\{X>50, X>30\}=P\{X>50\}.
$$

The survival probability is

$$
P\{X>t\}=e^{-0.03t}.
$$

Thus

$$
P\{X>50\mid X>30\}
=\frac{e^{-0.03\cdot 50}}{e^{-0.03\cdot 30}}
=e^{-0.6}.
$$

More generally,

$$
P\{X>t+s\mid X>t\}
=\frac{e^{-0.03(t+s)}}{e^{-0.03t}}
=e^{-0.03s}
=P\{X>s\}.
$$

This is the **memoryless property** of the exponential distribution.


```python
prob = np.exp(-0.6)
prob
```

```python
# Visualize the memoryless property by simulation.
rate = 0.03
X = rng.exponential(scale=1/rate, size=1_000_000)
empirical = (X[X > 30] > 50).mean()
theoretical = np.exp(-rate*20)
empirical, theoretical
```

# Summary: mental models

## Probability space

A probability space is a universe of possible outcomes plus a consistent size function on events.

## Event algebra

Events are sets. Probability problems often become set problems.

## Random variables

A random variable is not “random” by itself; it is a deterministic function of the random outcome.

## Distribution functions

The distribution function $F(b)=P\{X\le b\}$ compresses the law of a real-valued random variable into one nondecreasing, right-continuous function.

## Independence

Independence means joint probabilities factor into products of marginal probabilities.

## Conditional probability

Conditioning changes the universe to the event being conditioned on, then rescales probability.

## Bayes' formula

Bayes' formula reverses conditioning: from likelihoods $P(A\mid B_j)$ and priors $P(B_j)$ to posterior probabilities $P(B_j\mid A)$.
